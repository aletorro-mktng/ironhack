# CharacterTextSplitter with fixed chunk size

`CharacterTextSplitter` splits text into chunks close to a fixed character length. It tries to split on a separator first, so chunks may be smaller than `chunk_size` when a clean break is available.

```python
# If needed, install the package first:
# %pip install -q langchain-text-splitters

from langchain_text_splitters import CharacterTextSplitter
```

```python
text = """
LangChain provides several text splitters for preparing documents for retrieval.
CharacterTextSplitter is one of the simplest options. It measures chunks by
character count and splits using a separator such as a newline or space.

Fixed-size chunks are useful when you want predictable chunk lengths before
embedding text in a vector database. Chunk overlap can preserve context between
neighboring chunks.
""".strip()
```

```python
splitter = CharacterTextSplitter(
    separator="\n",
    chunk_size=100,
    chunk_overlap=20,
    length_function=len,
    is_separator_regex=False,
)

chunks = splitter.split_text(text)

for i, chunk in enumerate(chunks, start=1):
    print(f"Chunk {i} ({len(chunk)} characters):")
    print(chunk)
    print("-" * 40)
```

## Key parameters

- `chunk_size`: maximum target size for each chunk, measured by `length_function`.
- `chunk_overlap`: number of characters repeated between adjacent chunks.
- `separator`: preferred boundary for splitting text.
- `length_function`: usually `len`, but you can replace it with a token counter.
