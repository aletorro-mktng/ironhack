# Coding Lab Organization and Submission Steps

## STEP 1. Read the Lab Instructions

Before starting the lab, I read the instructions carefully.

Check whether the lab is asking me to submit:

- A GitHub repository link
- A Google Drive link
- Specific files
- Screenshots
- A Python script
- A Jupyter Notebook
- A `README.md`
- A `requirements.txt`
- Output files such as `.json`, `.csv`, or `.txt`

This helps decide how the lab should be organized and submitted.

## STEP 6 Install Required Packages

If the lab includes a requirements.txt file, open Terminal to install the packages with:

```bash
pip install -r requirements.txt
```

If there is no requirements.txt file, install the required packages manually.

I identify the needed packages by checking the lab instructions or the Python file for imported libraries. In some cases, VS Code may also suggest installing missing packages when you click Run, Play, or the triangle button.

Example of manually installing common packages:

```bash
pip install pandas numpy matplotlib scikit-learn openai python-dotenv
```
After installing the packages, run the lab again to make sure there are no missing package errors.

## STEP 7 Complete the Lab

Work through the lab instructions step by step.

Depending on the assignment, this may include:

* Writing Python code
* Completing a Jupyter Notebook
* Loading and analyzing a dataset
* Generating output files
* Creating charts
* Testing API calls
* Answering reflection questions
* Saving screenshots or results

## STEP 8 Test the Lab

Before submitting, run the lab to make sure everything works.

For a Python script, run:

```bash
python filename.py
```
Example:

```bash
python churn_prediction.py
```

For a Jupyter Notebook, run all cells from top to bottom and check for errors.

## STEP 9 Create a Requirements File if Needed

If the lab requires a list of installed packages, create a requirements.txt file.

Run:

```bash
pip freeze > requirements.txt
```
This lets someone else install the same packages later.

## STEP 10 Create or Update the README File

* Files included

* Notes about outputs or results

## 11. Add a `.gitignore` File

Before uploading to GitHub, create a `.gitignore` file to avoid uploading unnecessary or private files.

Example `.gitignore`:

text
venv/
.env
__pycache__/
.DS_Store
*.pyc


This prevents files like the virtual environment, API keys, and system files from being uploaded.

## 12. Create a GitHub Repository

se these steps when the lab instructions require a GitHub repository link.

### Step 1: Open the Lab Folder in VS Code

Open the completed lab folder in VS Code.

Make sure the folder includes the required files, such as:

- Python file or notebook
- Dataset, if allowed
- Output files
- Screenshots, if required
- `README.md`, if required
- `requirements.txt`, if required
- `.gitignore`

---

### Step 2: Check the Files Before Uploading

Before syncing to GitHub, check that the correct files are included.

Do not upload:

- `venv/`
- `.env`
- API keys
- Passwords
- `__pycache__/`
- `.DS_Store`

A basic `.gitignore` file can include:

```text
venv/
.env
__pycache__/
.DS_Store
*.pyc
```
## Step 3: Go to Source Control in VS Code

Click the Source Control icon on the left side of VS Code.

This shows all the files that were created or changed in the lab folder.

## Step 4: Stage the Files

Review the changed files.

Then stage the files by clicking the + icon next to each file or stage all changes if everything shown belongs to the lab.

## Step 5: Write a Commit Message

Write a short commit message that explains what you completed.

Examples:

Complete churn prediction lab
Add API product listing lab files
Update prompt engineering lab outputs

## Step 6: Commit the Changes

Click Commit in VS Code.

This saves the current version of the lab in Git.

## Step 7: Sync or Publish to GitHub

After committing, click Sync Changes.

If this is the first time uploading the lab folder to GitHub, VS Code may show Publish Branch instead of Sync Changes.

Click Publish Branch and choose GitHub as the remote location and make it public. 

VS Code will upload the lab folder to GitHub.

After the first upload, future updates can be pushed by staging, committing, and clicking Sync Changes.

## Step 8: Confirm the Upload

After syncing or publishing, open the GitHub repository link from VS Code or from your GitHub account.

Check that the correct files uploaded successfully.

Before sharing, open the Drive folder and confirm that all files uploaded correctly.

Change the sharing settings so the instructor can open the folder.

or using the specific sharing setting required by the instructor.
